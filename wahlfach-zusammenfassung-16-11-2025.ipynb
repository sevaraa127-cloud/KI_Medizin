```python
# Beispiel: Berechnung des BMI (Body-Mass-Index) eines Patienten
# Dieses Programm zeigt grundlegende Python-Funktionen:
# - Variablen zuweisen
# - Datentypen anzeigen
# - Mathematische Berechnungen durchführen
# - Einrückungen und print-Anweisungen
# - Kommentare verwenden

# Patientendaten (Variablenzuweisung)
name = "Anna Müller"      # String (Text)
gewicht_kg = 68.5         # Float (Kommazahl)
groesse_m = 1.70          # Float (Kommazahl)
alter = 32                # Integer (Ganzzahl)

# Ausgabe der Datentypen
print("Datentyp von 'gewicht_kg':", type(gewicht_kg))
print("Datentyp von 'name':", type(name))

# Mathematische Berechnung: BMI
bmi = gewicht_kg / (groesse_m ** 2)

# Kontrolle der BMI-Kategorie mit Einrückung (if-Struktur)
if bmi < 18.5:
    kategorie = "Untergewicht"
elif bmi < 25:
    kategorie = "Normalgewicht"
else:
    kategorie = "Übergewicht"

# Ausgabe der Ergebnisse
print("\n--- Patientenauswertung ---")
print("Name:", name)
print("Alter:", alter, "Jahre")
print("BMI:", round(bmi, 2))
print("Kategorie:", kategorie)
```

```python
# Ausgabe der Ergebnisse mit nur einem print-Statement

print("\n--- Patientenauswertung ---\n", "Name:", name, "\n", "Alter:", alter, "Jahre\n", "BMI:", round(bmi, 2), "\n", "Kategorie:", kategorie)
```

```python
print(
    f"""
--- Patientenauswertung ---
Name: {name}
Alter: {alter} Jahre
BMI: {round(bmi, 2)}
Kategorie: {kategorie}
"""
)
```

Wichtige Tools:

- Dictionary: Speichert Daten als Schlüssel–Wert-Paare (z. B. "name": "Laura").

- Indexing: Zugriff auf ein einzelnes Zeichen → name[0].

- Slicing: Teil eines Strings/Liste → diagnose[:3].

- Werte hinzufügen: patient["letzter_blutdruck"] = "130/85".

- Funktion: Wiederverwendbarer Codeblock → def ....

- return: Übergibt das Ergebnis zurück an die Stelle, an der die Funktion aufgerufen wird.

```python
# 1. Dictionary mit Patientendaten
patient = {
    "name": "Laura Becker",
    "alter": 45,
    "diagnose": "Hypertonie",
    "medikamente": ["Ramipril", "Amlodipin"]
}

# 2. Indexing – erster Buchstabe des Namens
erster_buchstabe = patient["name"][0]   # Zugriff auf das erste Zeichen

# 3. Slicing – erste drei Buchstaben der Diagnose
diagnose_kurz = patient["diagnose"][:3]  # Anfang bis Index 3 (ohne 3)

# 4. Neuen Wert zum Dictionary hinzufügen
patient["letzter_blutdruck"] = "130/85"   # Neues Schlüssel–Wert-Paar

# 5. Eigene Funktion mit return
def medikamenten_anzahl(patient_dict):
    """
    Gibt die Anzahl der Medikamente zurück.
    """
    return len(patient_dict["medikamente"])   # return gibt einen Wert zurück

anzahl_medikamente = medikamenten_anzahl(patient)

# 6. Ausgabe
print("--- Patientendaten ---")
print("Name:", patient["name"])
print("Alter:", patient["alter"])
print("Diagnose:", patient["diagnose"])
print("Diagnose (kurz):", diagnose_kurz)
print("Erster Buchstabe des Namens:", erster_buchstabe)
print("Medikamente:", patient["medikamente"])
print("Anzahl Medikamente:", anzahl_medikamente)
print("Letzter Blutdruck:", patient["letzter_blutdruck"])
```

##Machine Learning: Was ist das?

Machine Learning (ML) ist eine Methode, bei der Computer aus Daten lernen, anstatt dass wir ihnen jeden Schritt genau programmieren.

Stell dir vor, du bringst einem Hund bei, einen Ball zu holen:

- Du wirfst den Ball und belohnst den Hund, wenn er ihn zurückbringt.

- Nach vielen Versuchen lernt der Hund das Muster: „Ball → holen!“

ML funktioniert ähnlich:

- Der Computer bekommt viele Beispiele (Daten).

- Er lernt Muster in den Daten.

- Danach kann er Vorhersagen für neue Daten machen, die er noch nicht gesehen hat.

##Arten von Machine Learning

1. Überwachtes Lernen (Supervised Learning)

- Daten haben Labels / Beschriftungen.

    Beispiel: Bilder von Katzen und Hunden mit den Labels „Katze“ oder „Hund“.

- Der Computer lernt, neue Bilder richtig zu klassifizieren.

2. Unüberwachtes Lernen (Unsupervised Learning)

- Daten haben keine Labels.

- Der Computer versucht, selbst Muster oder Gruppen zu erkennen.

    Beispiel: Patienten mit ähnlichen Symptomen in Gruppen einteilen, ohne Diagnose zu kennen.

3. Bestärkendes Lernen (Reinforcement Learning)

- Der Computer lernt durch Versuch und Irrtum, mit Belohnungen oder Strafen.

    Beispiel: Roboter lernt laufen oder ein Programm lernt Schach spielen.

##K-Nearest Neighbors (KNN)

KNN ist ein Algorithmus, der Vorhersagen basierend auf Ähnlichkeit macht:

- Er schaut sich die k nächsten Punkte (Nachbarn) zu einem neuen Punkt an.

- Er schaut, welche Klasse bei diesen Nachbarn am häufigsten vorkommt.

- Dann sagt er: „Dieser neue Punkt gehört wahrscheinlich zu dieser Klasse.“

    Beispiel: Wir wollen vorhersagen, ob ein Patient Bluthochdruck (Hypertonie) hat, basierend auf Alter und Gewicht.

```python
# 1. Importieren der benötigten Bibliotheken
from sklearn.neighbors import KNeighborsClassifier  # KNN Algorithmus
import numpy as np  # für einfache Zahlenarrays

# Warum?
# sklearn bietet fertige ML-Algorithmen, numpy hilft uns, Daten als Arrays zu speichern.

# 2. Trainingsdaten erstellen
# Jede Zeile: [Alter, Gewicht]
X_train = np.array([
    [25, 60],
    [30, 70],
    [45, 80],
    [50, 90],
    [35, 65]
])

# Labels: 0 = kein Bluthochdruck, 1 = Bluthochdruck
y_train = np.array([0, 0, 1, 1, 0])

# Warum?
# X_train sind die Features, y_train die Labels. Das Modell lernt, welche Features zu welchem Label gehören.

# 3. KNN-Modell erstellen
knn = KNeighborsClassifier(n_neighbors=3)  # k = 3

# Warum?
# Wir wählen k=3, also schaut das Modell sich die 3 nächsten Nachbarn an.

# 4. Modell trainieren
knn.fit(X_train, y_train)

# Warum?
# Das Modell „merkt“ sich die Trainingsdaten. KNN speichert die Daten für späteren Vergleich.

# 5. Neue Patienten vorhersagen
# Beispiel: Patient ist 40 Jahre alt, 75 kg schwer
X_neu = np.array([[40, 75]])
vorhersage = knn.predict(X_neu)

# Warum?
# predict vergleicht den neuen Patienten mit den Trainingsdaten und gibt das wahrscheinlichste Label aus.

print("Vorhersage (0 = kein Hypertonie, 1 = Hypertonie):", vorhersage[0])
```

##Regression

Regression sagt einen kontinuierlichen Wert voraus, z. B. Blutdruck, Cholesterinwert oder Blutzucker.

Anders als KNN-Klassifikation, die Kategorien vorhersagt (0/1), sagt Regression eine Zahl vorher.

    Beispiel: Wir wollen den Blutdruck anhand von Alter und Gewicht vorhersagen.

```python
from sklearn.linear_model import LinearRegression  # Regressionsmodell
import numpy as np

# Trainingsdaten: [Alter, Gewicht]
X_train = np.array([[25, 60], [30, 70], [45, 80], [50, 90], [35, 65]])
# Zielvariable: Blutdruck systolisch in mmHg
y_train = np.array([120, 125, 140, 150, 130])

# 1. Regressionsmodell erstellen
modell = LinearRegression()

# 2. Modell trainieren
modell.fit(X_train, y_train)
# Erklärung: Das Modell lernt, wie Alter und Gewicht den Blutdruck beeinflussen

# 3. Vorhersage für neuen Patienten
X_neu = np.array([[40, 75]])
vorhersage = modell.predict(X_neu)
print("Vorhergesagter Blutdruck:", round(vorhersage[0], 1), "mmHg")

# 4. Score auf Trainingsdaten
genauigkeit = modell.score(X_train, y_train)
print("Score (R² auf Trainingsdaten):", round(genauigkeit, 2))
# Erklärung: R² gibt an, wie gut die Variablen Alter/Gewicht den Blutdruck erklären
```
