#Cross-Validation

Ziel: Überprüfen, wie gut ein Modell auf neuen, unbekannten Daten funktioniert.

1. Teile deine Daten in k gleich große Teile (z. B. 5 Teile → „5-Fold Cross-Validation“).

2. Trainiere das Modell auf k-1 Teilen und teste auf dem übrigen Teil.

3. Wiederhole das für jeden Teil.

4. Berechne die durchschnittliche Genauigkeit.

Beispiel:

Wir wollen den Cholesterinwert vorhersagen.

Cross-Validation hilft uns zu prüfen, ob unser Modell nicht nur die Trainingsdaten „auswendig gelernt“ hat, sondern auch neue Patienten korrekt vorhersagen kann.

```python
# 1. Importieren der Bibliotheken
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
import numpy as np

# 2. Trainingsdaten erstellen
# Features: [Alter, BMI]
X = np.array([
    [25, 22],
    [30, 25],
    [45, 28],
    [50, 30],
    [35, 27],
    [55, 32]
])

# Zielvariable: Cholesterinwerte in mg/dL
y = np.array([180, 190, 210, 230, 200, 240])

# 3. Lineares Regressionsmodell erstellen
modell = LinearRegression()

# 4. 5-Fold Cross-Validation durchführen
scores = cross_val_score(modell, X, y, cv=3)  # Standardmäßig R²
print("R² Scores für jedes Fold:", scores)
```

1. Import

- LinearRegression → für das Regressionsmodell

- cross_val_score → für Cross-Validation

- numpy → für einfache Arrays

2. Daten erstellen

- X = Features (Alter, BMI)

- y = Zielvariable (Cholesterin)

3. Modell erstellen

- LinearRegression() → erstellt ein lineares Regressionsmodell

4. Cross-Validation

- cross_val_score(modell, X, y, cv=5) → teilt die Daten in 5 Teile, trainiert auf 4 und testet auf 1, wiederholt für alle Teile

5. Ausgabe: R²-Score für jedes Fold und der Mittelwert
